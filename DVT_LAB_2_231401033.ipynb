## 1. Import Libraries

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import glob
import os

print("Libraries imported successfully.")
```

## 2. Download and Load the Kaggle Dataset

```python

# Install KaggleHub in Colab
!pip -q install kagglehub

import kagglehub

# Download the REAL Kaggle dataset
dataset_path = kagglehub.dataset_download(
    "adrianjuliusaluoch/daily-google-search-trends-us"
)

# Locate the CSV automatically
csv_files = glob.glob(os.path.join(dataset_path, "**", "*.csv"), recursive=True)

print("Kaggle dataset downloaded successfully.")
print("Dataset folder:", dataset_path)
print("CSV files found:", csv_files)

csv_path = csv_files[0]
df = pd.read_csv(csv_path)

print("\nDataFrame loaded successfully.")
print("Number of records:", len(df))
print("Number of columns:", len(df.columns))
```

## 14. Filter the DataFrame

```python

# Demonstrate filtering using the first column.
# For text columns, keep non-null values.
# For numeric columns, keep values greater than or equal to the median.

first_col = df.columns[0]

if pd.api.types.is_numeric_dtype(df[first_col]):
    condition = df[first_col] >= df[first_col].median()
else:
    condition = df[first_col].notna()

filtered_df = df[condition]

print("Filtering condition applied on:", first_col)
print("Filtered rows:", len(filtered_df))
display(filtered_df.head())
```

## 15. Check for Missing Values

```python

missing = df.isnull().sum()

print("Missing values in each column:")
display(missing)

print("Total missing values:", missing.sum())
```

## 16. Check for Duplicate Records

```python

duplicates = df.duplicated().sum()

print("Number of duplicate records:", duplicates)
```

## 17. Value Counts

```python

# Choose the first categorical/text column
text_columns = df.select_dtypes(include="object").columns

if len(text_columns) > 0:
    value_col = text_columns[0]
    print("Value frequency for:", value_col)
    display(df[value_col].value_counts(dropna=False).head(10))
else:
    print("No text/categorical column was found.")
```

## 18. Simple Visualization

```python

# Plot the first numeric column if available
numeric_columns = df.select_dtypes(include=np.number).columns

if len(numeric_columns) > 0:
    num_col = numeric_columns[0]

    plt.figure(figsize=(10, 5))
    plt.hist(df[num_col].dropna(), bins=20)
    plt.title("Distribution of " + num_col)
    plt.xlabel(num_col)
    plt.ylabel("Frequency")
    plt.show()
else:
    print("No numeric column was found for histogram visualization.")
```
